```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", pd.NA)
```

```python
df[df["TotalCharges"].isna()]
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```
